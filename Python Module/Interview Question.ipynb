```python
# Find the second largest number
n = [23,433,89,90,15,92,20]
n.sort()
print(n[-2])

#using max and remove
n.remove(max(n))
print(max(n))

#without using built-in function
largest = n[0]
slargest = n[0]
for i in range(len(n)):
    if n[i] > slargest and n[i] != largest:
        slargest = n[i]
print(slargest)
```

```python
# remove duplicate string
s = "sssssfffddddddjjhsss"
l = list(set(s))
a = ""
for i in l:
    a = a+i 
print(a) # this sol will disturb the order of characters bcz we are using set

# using list comprehension
a = ""
[a := a+i for i in s if i not in a]
print(a)
# := "It is an assignment expression operator (walrus operator) that allows assignment inside expressions to reduce redundant computations."

# without using built-in function
a = ""
for i in s:
    if i not in a:
        a +=i
print(a)
```

```python
# Reverse a string
s = "Nshkkkp"
print(s[::-1])
# or
ns = reversed(s)
print(' '.join(ns))

#wihtout built-in func
res = ''
for i in range(len(s)-1,-1,-1): #step is -1 bcz we are moving in backword way
    res += s[i]
print(res)
```

```python
#Find all duplicates in a List
l = [1,2,2,3,4,8,9,2,3]
dup = []
for i in l:
    if l.count(i)>1:
        dup.append(i)
print(set(dup))

#using counter 
from collections import Counter
ll = dict(Counter(l))
ab = []
for v in ll.values():
    if v > 1:
        ab.append(ll.key(v))
```
