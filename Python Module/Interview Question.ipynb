```python
#Find all duplicates in a List
l = [1,2,2,3,4,8,9,2,3]
dup = []
for i in l:
    if l.count(i)>1:
        dup.append(i)
print(set(dup))

#using counter 
from collections import Counter
l = [1,2,3,2,4,5,3,6]
ll = dict(Counter(l))
ab = []
for key, v in ll.items():
    if v > 1:
        ab.append(key)
print(ab)

#ALTENATE
l = [1,2,2,3,4,8,9,2,3]
a = {}
for i in l:
    if i not in a:
        a[i] = 1
    else:
        a[i] = a[i]+1
print(a)

b = []
for key,val in a.items():
    if val > 1:
        b.append(key)
print(b)
```

```python
#Find most frequent element in the list
l = [1,2,2,3,4,8,9,2,3]
from collections import Counter
freq = dict(Counter(l))
count = 0
ele = None
for k,v in freq.items():
    if v > count:
        count = v
        ele = k
print(ele)
```

```python
# create a account as class, deposite 500, withdraw 300 then withdraw 1200 if there is insufficient balance print  insufficient funds then at last get balance
class BankAccount:
    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance
    
    def deposite(self, value):
        self.balance = self.balance + value
        print(f"[Transaction] deposite of {value} | Balance: {self.balance}")

    def withdraw(self, value):
        if value > self.balance:
            print("insufficient funds")
        else:
            self.balance = self.balance - value
            print(f"[Transaction] withdraw of {value} | Balance: {self.balance}")

    def get_balance(self):
        print(self.balance)

#create account
acc = BankAccount("Naina",1000)

#Deposit
acc.deposite(500)

#Withdraw
acc.withdraw(300)
acc.withdraw(2000)

#get balance
acc.get_balance()
```

```python
s = "2[a]3[bd]"
# expected o/p:- aabdbdbd
import re
parts = re.findall(r'(\d+)\[([a-zA-Z]+)\]',s)
result = "".join(int(num)* text for num, text in parts)
print(result)

#Without Built-in function
s = "2[a]3[bd]"
result = ""
num = 0
temp = ""
for ch in s:
    if ch.isdigit():
        num = num*10 + int(ch)   # handles multi-digit numbers
        
    elif ch == '[':
        temp = ""
        
    elif ch == ']':
        result += temp * num
        num = 0
        
    else:
        temp += ch
print(result)
```

```python
# Given a string: # "machine number : 345 data number : 435 calc: 100 calc:200"
# Extract machine number, data number and all calc values using Python.
line = "machine number : 345 data number : 435 calc: 100 calc:200"
import re
line = "machine number : 345 data number : 435 calc: 100 calc:200"
machine_num = re.search(r"machine number\s*:\s*(\d+)", line).group(1)
data_num = re.search(r"data number\s*:\s*(\d+)", line).group(1)
calc_vals = re.findall(r"calc\s*:\s*(\d+)", line)
print("Machine:", machine_num)
print("Data:", data_num)
print("Calc:", calc_vals)
```
