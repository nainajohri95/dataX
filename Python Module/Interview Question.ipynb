```python
# Find the second largest number
n = [23,433,89,90,15,92,20]
n.sort()
print(n[-2])

#using max and remove
n.remove(max(n))
print(max(n))

#without using built-in function
largest = n[0]
slargest = n[0]
for i in range(len(n)):
    if n[i] > slargest and n[i] != largest:
        slargest = n[i]
print(slargest)
```

```python
# remove duplicate string
s = "sssssfffddddddjjhsss"
l = list(set(s))
a = ""
for i in l:
    a = a+i 
print(a) # this sol will disturb the order of characters bcz we are using set

# using list comprehension
a = ""
[a := a+i for i in s if i not in a]
print(a)
# := "It is an assignment expression operator (walrus operator) that allows assignment inside expressions to reduce redundant computations."

# without using built-in function
a = ""
for i in s:
    if i not in a:
        a +=i
print(a)
```

```python
# Reverse a string
s = "Nshkkkp"
print(s[::-1])
# or
ns = reversed(s)
print(' '.join(ns))

#wihtout built-in func
res = ''
for i in range(len(s)-1,-1,-1): #step is -1 bcz we are moving in backword way
    res += s[i]
print(res)
```

```python
#Find all duplicates in a List
l = [1,2,2,3,4,8,9,2,3]
dup = []
for i in l:
    if l.count(i)>1:
        dup.append(i)
print(set(dup))

#using counter 
from collections import Counter
l = [1,2,3,2,4,5,3,6]
ll = dict(Counter(l))
ab = []
for key, v in ll.items():
    if v > 1:
        ab.append(key)
print(ab)
```

```python
#Find most frequent element in the list
l = [1,2,2,3,4,8,9,2,3]
from collections import Counter
freq = dict(Counter(l))
count = 0
ele = None
for k,v in freq.items():
    if v > count:
        count = v
        ele = k
print(ele)
```

```python
# Merge two lists and remove duplicates while keeping order
list1 = [10,20,50,40]
list2 = [50,60,70,10]
l = []
for i in list1+list2:
    if i not in l:
        l.append(i)
print(l)
```

```python
#Factorial of 5
def fact(n):
    if n == 0:
        return 1
    else:
        return n*fact(n-1)
fact(5)
```

```python
# create a account as class, deposite 500, withdraw 300 then withdraw 1200 if there is insufficient balance print  insufficient funds then at last get balance
class BankAccount:
    def __init__(self, name, balance=0):
        self.name = name,
        self.balance = balance
    
    def deposite(self, value):
        balance = balance+value
        print(f"[Transaction] deposite of {value} | Balance: {self.balance}")

    def withdraw(self, value):
        balance = balance - value
        print(f"[Transaction] deposite of {value} | Balance: {self.balance}")

    def get_balance(self):
        print(balance)

#create account
acc = BankAccount("Naina",1000)

#Deposit
acc.deposite(500)

#Withdraw
acc.withdraw(300)
acc.withdraw(2000)

#get balance
acc.get_balance()
```
